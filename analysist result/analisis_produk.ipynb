**KOLOM NAMA PRODUK**

```python
nama_normal = df_produk['Nama_Produk'].str.strip().str.lower()
variasi = df_produk.groupby(nama_normal)['Nama_Produk'].nunique()
variasi[variasi > 1]
```

```python
nama_produk = df_produk['Nama_Produk'].str.strip().str.lower()

for nama1, nama2 in combinations(nama_produk, 2):
    similarity = SequenceMatcher(None, nama1, nama2).ratio()
    if similarity >= 0.8:
        print(nama1, '<->', nama2, '=', round(similarity * 100, 2), '%')
```

**KOLOM KATEGORI**

```python
nama_normal = df_produk['Kategori'].str.strip().str.lower()

variasi = df_produk.groupby(nama_normal)['Kategori'].nunique()
variasi[variasi > 1]
```

```python
nama_produk = df_produk['Kategori'].str.strip().str.lower()

for nama1, nama2 in combinations(nama_produk, 2):
    similarity = SequenceMatcher(None, nama1, nama2).ratio()
    if similarity >= 0.8:
        print(nama1, '<->', nama2, '=', round(similarity * 100, 2), '%')
```
